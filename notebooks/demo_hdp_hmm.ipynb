# Demo: HDP-HMM truncado

Este notebook ejecuta la aproximacion truncada del iHMM basada en HDP-HMM sobre una partitura de ejemplo.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent
elif not (PROJECT_ROOT / 'src').exists() and (PROJECT_ROOT.parent / 'src').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

EXAMPLE_SCORE = PROJECT_ROOT / 'examples' / 'example_score.musicxml'
OUTPUT_DIR = PROJECT_ROOT / 'outputs' / 'notebook_hdp'

print('PROJECT_ROOT =', PROJECT_ROOT)
print('EXAMPLE_SCORE =', EXAMPLE_SCORE)
```

```python
from examples.generate_example_score import main as generate_example_score

generate_example_score()
```

```python
from src.data.parsing import parse_score, extract_events
from src.data.observations import build_observation_sequence
from src.models.hdp_hmm import TruncatedHDPHMM
from src.analysis.diagnostics import summarize_hdp_result
from src.analysis.interpretation import interpret_hdp_states

score = parse_score(EXAMPLE_SCORE)
events = extract_events(score)
observations = build_observation_sequence(events, 'pitch_class')

model = TruncatedHDPHMM(n_states=20, n_iters=120, burn_in=60, seed=7)
result = model.fit(observations)
summarize_hdp_result(result)
```

```python
interpret_hdp_states(result)
```

```python
from src.analysis.visualization import generate_hdp_figures

generate_hdp_figures(result, OUTPUT_DIR)
```
